```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('spam.csv',encoding='latin1')
```

```python
df
```

```python
df=df.drop(columns=['Unnamed: 2','Unnamed: 3','Unnamed: 4'])
```

```python
df=df.drop_duplicates()
```

```python
df['v2']=df['v2'].apply(lambda x: x.lower())
df['v2']=df['v2'].apply(lambda x:x.replace(',',''))
```

```python
df.reset_index().drop(columns=['index'],inplace=True)
```

```python
from sklearn.feature_extraction.text import CountVectorizer 
```

```python
words=CountVectorizer(max_features=400,stop_words='english')
```

```python
words.fit_transform(df['v2'][df['v1']=='spam'])
```

```python
words.get_feature_names_out()
```

```python
df['v2']=df['v2'].apply(lambda x: x.split())
```

```python
#model runnning
def detector(input):
    count=0
    for character in  input :
        if character in words.get_feature_names_out():
            count+=1
            if count==5:
                return 'spam'
    return 'ham'
 
        
```

```python
df
```

```python
df['detected']=df['v2'].apply(lambda x: detector(x))
```
